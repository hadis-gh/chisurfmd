```python
import matplotlib.pyplot as plt
import numpy as np


position = np.loadtxt("atom_path.txt")

x = position[:, 0]
y = position[:, 1]

plt.figure(figsize=(8,8))

plt.scatter(x, y, s=30 , c=position[:,2])
plt.title("Atom diffusion path")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(x[0]-1, x[-1]+1)
plt.ylim(y[0]-1, y[-1]+1)
plt.grid(True)
plt.show()
```
