## Filter E(R) with R, E ranges (R opt plot)

Select R and E range -> $\phi_1, \phi_2, \zeta$ 

```python
import numpy as np
import matplotlib.pyplot as plt

file = '../E_all_DS.dat'

data = np.loadtxt(file)

R_interest = 8.8
R_precision = 0.01
energy_interest = -0.8
energy_precision = 0.0001

ref_energy = -6227.1749
energy_scale = 2 * ref_energy
R_range = [R_interest - R_precision, R_interest + R_precision]
energy_range = [(energy_interest - energy_precision) + energy_scale,
                energy_interest + energy_precision + energy_scale]

mask = (data[:, 3] >= R_range[0]) & (data[:, 3] <= R_range[1]) & (data[:, 4] >= energy_range[0]) & (data[:, 4] <= energy_range[1])
extracted_data = data[mask]

print(f"Found {extracted_data.shape[0]} rows")
print("phi1  phi2     zeta       R      E")
print("-" * 40)
for row in extracted_data:
    phi1, phi2, zeta, R, E = row
    print(f"{phi1:4.0f}  {phi2:4.0f}  {zeta:8.6f}  {R:5.1f}  {E:10.4f}")
```
