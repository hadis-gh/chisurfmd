```python
import numpy as np
import matplotlib.pyplot as plt

def draw_circles(length, circles, color="black", figax=None):
    if figax is None:
        fig, ax = plt.subplots()
    else:
        fig, ax = figax

    ax.set_xlim(0, length)
    ax.set_ylim(0, length)

    for c in circles:
        if isinstance(c, np.ndarray) and len(c) >= 3:
            circle = plt.Circle(c[:2], c[2], fill=False, color=color)
            ax.add_artist(circle)
        elif isinstance(c, np.ndarray) and len(c) == 2:
            circle = plt.Circle(c[:2], 1, fill=False, color=color)  # Default radius for single point
            ax.add_artist(circle)

    ax.set_aspect('equal', adjustable='box')
    return fig, ax

def converter_func(s):
    return float(s.decode('utf-8').strip('()'))

def read_centers_numpy(file_name):
    return np.loadtxt(file_name, delimiter=',', converters={0: converter_func, 1: converter_func})

length = 100
file_name = '/home/hadis/custom_vector/build/test/circle3Lines{}.txt'
circles = read_centers_numpy(file_name.format(""))

figax = draw_circles(length, circles)

movingCircles = read_centers_numpy(file_name.format("Moving"))

# movingCircles[-1,2] = 2     #changing the radius to magnify the moving circle
fig, ax = draw_circles(length, movingCircles, "red", figax)

# ax.plot(np.arange(0,100))
```
