```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches

originPoint= (20 , 20)
startPoint= (20 , 20)
movementDirection= (1 , 3)
length= 50
circlesRadius= 1

intersection= (21.7071 , 25.1214)
NumberOfIteration= 0

circlesCenters= [(17.3268 , 16.8371), (22.3821 , 25.8593), (29.5044 , 1.32919), (12.6894 , 31.6499), (10.0439 , 23.4122), (23.9932 , 0.158378), (14.1566 , 15.6929), (10.6359 , 13.5155), (13.3703 , 46.3044), (22.1481 , 38.9573), (47.6281 , 27.722), (2.26899 , 27.2219), (14.2514 , 23.6402), (26.4509 , 26.2434), (6.12027 , 40.5263), (47.1566 , 7.56528), (17.3663 , 7.27126), (7.70616 , 8.5315), (49.3487 , 21.0256), (21.9225 , 9.0384), (28.9846 , 3.8113), (40.3396 , 19.0932), (40.743 , 0.92829), (36.0645 , 28.2741), (23.4244 , 19.622), (23.9932 , 50.1584), (-0.651279 , 21.0256), (40.743 , 50.9283), ]

slope = movementDirection[1] / movementDirection[0]
x_range = np.linspace(originPoint[0], originPoint[0] + 100 * movementDirection[0], 100)
y_range = originPoint[0] + slope * (x_range - originPoint[0])

fig, ax = plt.subplots()

ax.add_patch(patches.Rectangle((0, 0), length, length, fill=False))

ax.plot(*originPoint, 'go')
endPoint = (originPoint[0] + movementDirection[0], originPoint[1] + movementDirection[1])

if originPoint==startPoint:
    ax.plot([originPoint[0], intersection[0]], [originPoint[1], intersection[1]], 'r-')
else:
    ax.plot(x_range, y_range, '-r')

for center in circlesCenters:
    circle = patches.Circle(center, circlesRadius, fill=False)
    ax.add_patch(circle)

ax.plot(*intersection, 'bx')

ax.set_xlim(0, length)
ax.set_ylim(0, length)

ax.set_aspect('equal', adjustable='box')

plt.show()
```
