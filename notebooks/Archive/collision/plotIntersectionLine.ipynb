```python
import numpy as np
import matplotlib.pyplot as plt

def plot_line_and_circle(origin_x, origin_y, circle_center_x, circle_center_y, radius, intersection_x, intersection_y):
    fig, ax = plt.subplots(figsize=(10, 10))

    circle = plt.Circle((circle_center_x, circle_center_y), radius, fill=False)
    ax.add_artist(circle)

    line_x = [origin_x, intersection_x]
    line_y = [origin_y, intersection_y]
    ax.plot(line_x, line_y, color='r', label='Line')

    ax.plot(origin_x, origin_y, 'bo', label='Origin')

    ax.plot(intersection_x, intersection_y, 'ro', label='Intersection')

    ax.axvline(x=0, linestyle='--', linewidth=1, label='x=0')
    ax.axhline(y=0, linestyle='--', linewidth=1, label='y=0')

    ax.set_xlim(min(origin_x, circle_center_x - radius, intersection_x) - 1,
                max(origin_x, circle_center_x + radius, intersection_x) + 1)
    ax.set_ylim(min(origin_y, circle_center_y - radius, intersection_y) - 1,
                max(origin_y, circle_center_y + radius, intersection_y) + 1)

    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_title('Line and Circle')
    ax.legend()
    ax.set_aspect('equal', adjustable='box')

    plt.show()

plot_line_and_circle(0, 0,
                     3, -1, 1, 
                     3, 0)
```
