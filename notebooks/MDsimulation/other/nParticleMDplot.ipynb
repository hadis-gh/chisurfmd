```python
import numpy as np
import matplotlib.pyplot as plt

energyK_euler = np.loadtxt("/home/hadis/custom_vector/build/test/euler/N_particle_KineticEnergyMD.dat")
energyK_vv = np.loadtxt("/home/hadis/custom_vector/build/test/velocityVerlet/N_particle_KineticEnergyMD.dat")

total_KE_euler = np.sum(energyK_euler, axis = 1)
total_KE_vv = np.sum(energyK_vv, axis = 1)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 10))

for i in range(energyK_vv.shape[1]):
    ax1.plot(energyK_euler[:, i], label=f'Particle {i+1}')

ax1.set_xlabel('Time')
ax1.set_ylabel('Energy')
ax1.set_title('Kinetic Energy of Each Particle')
ax1.legend()

ax2.plot(total_KE_vv, label=f'velocity verlet')
ax2.plot(total_KE_euler, label=f'euler')
ax2.set_xlabel('Time')
ax2.set_ylabel('Energy')
ax2.set_title('Kinetic Energy of system')
ax1.legend()

plt.tight_layout()
plt.show()
```
