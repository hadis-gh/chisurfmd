```python
import numpy as np
import matplotlib.pyplot as plt

# plt.style.use(['science', 'notebook', 'grid'])

sigma = 1
epsilon = 1

def ljPotential(r):
    return 4 * epsilon * (sigma ** 12)/(r ** 12) - 4 * epsilon * (sigma ** 6)/(r ** 6)

def ljForce(r):
    return 48 * epsilon * (sigma ** 12) / (r **13) - 24 * epsilon * (sigma ** 6)/ (r**7)

r = np.linspace(0.9 * sigma, 3 * sigma, 400)

potential = ljPotential(r)
force = ljForce(r)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.plot(r, potential, label='Potential')
plt.title('Lennard-Jones Potential')
plt.xlabel('Distance (r)')
plt.ylabel('Potential Energy (U)')
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)
plt.plot(r, force, label='Force', color='orange')
plt.title('Lennard-Jones Force')
plt.xlabel('Distance (r)')
plt.ylabel('Force (F)')
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

class LjOrientedPotential:
    def __init__(self, phiConst = 1, sigma = 1, epsilon = 1):
        self.phiConst = phiConst
        self.epsilon = epsilon
        self.sigma6 = sigma**6
        self.sigma12 = self.sigma6*self.sigma6
    def __call__(self, r, deltaPhi):
        r6 = r**6
        r12 = r6 * r6
        return 4.0 * self.epsilon * (self.sigma12 / r12 - 0.5 * self.sigma6 / r6) * self.phiConst * np.cos(self.phiConst * deltaPhi)
    
pot = LjOrientedPotential(phiConst=4)

L = 100
grid = np.meshgrid(np.linspace(1.1, 3, L), np.linspace(0, np.pi, L))

E = pot(grid[0], grid[1])

plt.imshow(E, extent=[grid[0][0][0], grid[0][-1][-1], grid[1][0][0], grid[1][-1][-1]])
plt.xlabel("r")
plt.ylabel("$\\Delta\\phi$")
plt.colorbar()
plt.tight_layout()
# plt.savefig("orientedPot.png", bbox_inches='tight', 
#                transparent=True,
#                pad_inches=0)
```

```python
import numpy as np
import matplotlib.pyplot as plt

class Ljo_potential:
    def __init__(self, phi_const):
        self.phi_const = phi_const
    def __call__(self, r, deltaPhi):
        r6 = r ** 6
        r12 = r ** 12
        return 4.0 * (1 / r12 - 0.5 / r6) * self.phi_const * np.cos(self.phi_const * deltaPhi)
    
potential= Ljo_potential(phi_const=4)
L = 100

r = np.linspace(1.1, 3, L)
delta_phi = np.linspace(0, np.pi, L)

R, Delta_phi= np.meshgrid(r, delta_phi)

potential_values = potential(R, Delta_phi)

fig, axs = plt.subplots(1, 2, figsize=(10, 6), dpi=100)
im = axs[0].imshow(potential_values,  aspect='auto')
axs[0].set_xlabel('r')
axs[0].set_ylabel('$\\Delta\\phi$')
fig.colorbar(im, label='Potential')

c= axs[1].contourf(R, Delta_phi, potential_values, levels=30)
axs[1].set_xlabel("r")
plt.tight_layout()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

class LjPotential:
    def __call__(self, r):
        r6 = r ** 6
        r12 = r ** 12
        return 4.0 * (1 / r12 - 1 / r6)
    
class MultiplicativePotential:
    def __init__(self, periodicity):
        self.periodicity = periodicity
    def __call__(self, r, deltaPhi):
        r6 = r ** 6
        r12 = r ** 12
        lj_pot = LjPotential()
        return lj_pot(r) * np.cos(self.periodicity * deltaPhi)
    
class AdditivePotential:
    def __init__(self, periodicity):
        self.periodicity = periodicity
    def __call__(self, r, deltaPhi):
        r6 = r ** 6
        r12 = r ** 12
        lj_pot = LjPotential()
        return lj_pot(r) - 20 * (r**-6) * np.cos(self.periodicity * deltaPhi)
    

multy_potential= MultiplicativePotential(periodicity=4)
addictive_potential= AdditivePotential(periodicity=2)

L = 100

r = np.linspace(0.9, 1.3, L)
delta_phi = np.linspace(0, 2*np.pi, L) ##pi or 2*pi (pi can be 0 to 2pi >> deltaPi: ...) ?? How to show in degree?

R, Delta_phi= np.meshgrid(r, delta_phi)

potential_values_add = addictive_potential(R, Delta_phi)
potential_values_multi = multy_potential(R, Delta_phi)

fig, axs = plt.subplots(1, 2, figsize=(10, 6), dpi=100)
c= axs[0].contourf(R, Delta_phi, potential_values_add, levels=100)
# fig.colorbar(c, label='Potential')
axs[0].set_xlabel('r')
axs[0].set_ylabel('$\\Delta\\phi$')
axs[0].set_title('$V_LJ - A(r)cos(n \Delta\Phi)$')
# axs[0].set_xlim()

c= axs[1].contourf(R, Delta_phi, potential_values_multi, levels=100)
fig.colorbar(c, label='Potential')
axs[1].set_xlabel("r")
axs[1].set_title('$V_L * cos(n \Delta\Phi)$')
plt.tight_layout()

plt.show()
```
