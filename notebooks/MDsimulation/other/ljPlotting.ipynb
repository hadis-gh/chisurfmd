```python
import numpy as np
import matplotlib.pyplot as plt

# plt.style.use(['science', 'notebook', 'grid'])

sigma = 1
epsilon = 1

def ljPotential(r):
    return 4 * epsilon * (sigma ** 12)/(r ** 12) - 4 * epsilon * (sigma ** 6)/(r ** 6)

def ljForce(r):
    return 48 * epsilon * (sigma ** 12) / (r **13) - 24 * epsilon * (sigma ** 6)/ (r**7)

r = np.linspace(0.9 * sigma, 3 * sigma, 400)

potential = ljPotential(r)
force = ljForce(r)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.plot(r, potential, label='Potential')
plt.title('Lennard-Jones Potential')
plt.xlabel('Distance (r)')
plt.ylabel('Potential Energy (U)')
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)
plt.plot(r, force, label='Force', color='orange')
plt.title('Lennard-Jones Force')
plt.xlabel('Distance (r)')
plt.ylabel('Force (F)')
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

class LjOrientedPotential:
    def __init__(self, phiConst = 1, sigma = 1, epsilon = 1):
        self.phiConst = phiConst
        self.epsilon = epsilon
        self.sigma6 = sigma**6
        self.sigma12 = self.sigma6*self.sigma6
    def __call__(self, r, deltaPhi):
        r6 = r**6
        r12 = r6 * r6
        return 4.0 * self.epsilon * (self.sigma12 / r12 - 0.5 * self.sigma6 / r6) - self.phiConst * np.cos(self.phiConst * deltaPhi)
    
pot = LjOrientedPotential(phiConst=1)

L = 100
grid = np.meshgrid(np.linspace(1.1, 3, L), np.linspace(0, np.pi, L))

E = pot(grid[0], grid[1])

plt.imshow(E, extent=[grid[0][0][0], grid[0][-1][-1], grid[1][0][0], grid[1][-1][-1]])
plt.xlabel("r")
plt.ylabel("$\\Delta\\phi$")
plt.tight_layout()
plt.savefig("orientedPot.png", bbox_inches='tight', 
               transparent=True,
               pad_inches=0)

grid[0][-1][-1]
```

```python
import numpy as np
import matplotlib.pyplot as plt

class Ljo_potential:
    def __init__(self, phi_const):
        self.phi_const = phi_const
    def __call__(self, r, deltaPhi):
        r6 = r ** 6
        r12 = r ** 12
        return 4.0 * (1 / r12 - 0.5 / r6) - self.phi_const * np.cos(self.phi_const * deltaPhi)
    
potential= Ljo_potential(phi_const=1)
L = 100

r = np.linspace(1, 10, L)
delta_phi = np.linspace(0, np.pi, L)

R, Delta_phi= np.meshgrid(r, delta_phi)
potential_values = potential(R, Delta_phi)

fig, ax = plt.subplots()
im = ax.imshow(potential_values, extent=(1, 10, 0, 180), origin='lower', aspect='auto')
ax.set_xlabel('r')
ax.set_ylabel('deltaPhi')
fig.colorbar(im, ax=ax, label='Potential')
plt.show()
```
