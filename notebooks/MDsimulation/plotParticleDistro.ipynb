```python
import numpy as np
import matplotlib.pyplot as plt

def draw_circles(radius, length, centers):
    fig, ax = plt.subplots()

    ax.set_xlim(0, length)
    ax.set_ylim(0, length)

    for center in centers:
        circle = plt.Circle(center, radius, fill=False)
        ax.add_artist(circle)

    ax.set_aspect('equal', adjustable='box')
    plt.show()

def converter_func(s):
    return float(s.decode('utf-8').strip('()'))

def read_centers_numpy(file_name):
    return np.loadtxt(file_name, delimiter=',', converters={0: converter_func, 1: converter_func})

circle_radius = 1
length = 50

file_name = '/home/hadis/custom_vector/build/test/CirclesDistr.txt'
circle_centers = read_centers_numpy(file_name)

draw_circles(circle_radius, length, circle_centers)
```
