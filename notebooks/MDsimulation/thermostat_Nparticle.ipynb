```python
# path of particles for Symplectic Euler VS Velocity Verlet method

import numpy as np
import matplotlib.pyplot as plt

def readPositions(filename):
    data = np.loadtxt(filename)
    numSteps = data.shape[0]
    numParticles = data.shape[1] // 2
    positions = data.reshape(numSteps, numParticles, 2)
    return positions

def plotTraj(inputs, labels):

    fig, axs = plt.subplots(len(inputs), 2, figsize=(10, 15))

    for j, fname in enumerate(inputs):
        label = labels[j]
        pos = readPositions(fname)

        axs[j][0].scatter(pos[0, :, 0], pos[0, :, 1], label='Initial Positions', color='blue')
        axs[j][0].scatter(pos[-1, :, 0], pos[-1, :, 1], label='Final Positions', marker='x', color='red')
        axs[j][0].set_title('Initial and final positions', fontsize = 15)
        axs[j][0].set_ylabel(label, fontsize = 15)
        axs[j][0].legend()

        for i in range(pos.shape[1]):
            axs[j][1].plot(pos[:, i, 0], pos[:, i, 1], lw=.5)
        axs[j][1].set_title(label, fontsize = 15)
```

```python
inputs = ["/home/hadis/custom_vector/build/test/velocityVerlet/N_particle_PosMD.dat",
    "/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/velocityScaling/N_particle_PosMD.dat",
    "/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/berendsen/N_particle_PosMD.dat",
    "/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/andersen/N_particle_PosMD.dat"
]
labels = ['Without Thermostat',
'particles path(VV)', 'Velocity Scaling Thermostat(VV)', 'Berendsen Thermostat(VV)','Andersen Thermostat(VV)'
]

plotTraj(inputs, labels)
```

```python
# energy of system for thermostats

import numpy as np
import matplotlib.pyplot as plt

kinetic_none = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/N_particle_KineticEnergyMD.dat'), axis = 1)
potential_none = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/N_particle_PotentialEnergyMD.dat'), axis = 1)
total_none = kinetic_none + potential_none

kinetic_vs = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/velocityScaling/N_particle_KineticEnergyMD.dat'), axis = 1)
potential_vs = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/velocityScaling/N_particle_PotentialEnergyMD.dat'), axis = 1)
total_vs = kinetic_vs + potential_vs

kinetic_br = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/berendsen/N_particle_KineticEnergyMD.dat'), axis = 1)
potential_br = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/berendsen/N_particle_PotentialEnergyMD.dat'), axis = 1)
total_br = kinetic_br + potential_br

kinetic_an = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/andersen/N_particle_KineticEnergyMD.dat'), axis = 1)
potential_an = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/andersen/N_particle_PotentialEnergyMD.dat'), axis = 1)
total_an = kinetic_an + potential_an

fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 18))

ax1.plot(kinetic_none, label=f'no thermostat')
ax1.plot(kinetic_vs, label=f'velocity scaling thermostat')
ax1.plot(kinetic_br, label=f'berendsen thermostat')
ax1.plot(kinetic_an, label=f'andersen thermostat')
ax1.set_xlabel('Time')
ax1.set_ylabel('Energy')
ax1.set_title('Kinetic Energy of system', fontsize = 15)
ax1.legend()
ax1.grid()

ax2.plot(potential_none, label=f'no thermostat')
ax2.plot(potential_vs, label=f'velocity scaling thermostat')
ax2.plot(potential_br, label=f'berendsen thermostat')
ax2.plot(potential_an, label=f'andersen thermostat')
ax2.set_xlabel('Time')
ax2.set_ylabel('Energy')
ax2.set_title('Potential Energy of system', fontsize = 15)
ax2.legend()
ax2.grid()

ax3.plot(total_none, label=f'no thermostat')
ax3.plot(total_vs, label=f'velocity scaling thermostat')
ax3.plot(total_br, label=f'berendsen thermostat')
ax3.plot(total_an, label=f'andersen thermostat')
ax3.set_xlabel('Time')
ax3.set_ylabel('Energy')
ax3.set_title('Total Energy of system', fontsize = 15)
ax3.legend()
ax3.grid()
```
