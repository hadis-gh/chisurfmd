```python
# path of particles for Symplectic Euler VS Velocity Verlet method

import numpy as np
import matplotlib.pyplot as plt

def readPositions(filename):
    data = np.loadtxt(filename)
    numSteps = data.shape[0]
    numParticles = data.shape[1] // 2
    positions = data.reshape(numSteps, numParticles, 2)
    return positions

path_Wthermo = "/home/hadis/custom_vector/build/test/velocityVerlet/N_particle_PosMD.dat"
pos_Wthermo = readPositions(path_Wthermo)

path_VSthermo = "/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/velocityScaling/N_particle_PosMD.dat"
pos_VSthermo = readPositions(path_VSthermo)

path_Bthermo = "/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/berendensen/N_particle_PosMD.dat"
pos_Bthermo = readPositions(path_Bthermo)

fig, axs = plt.subplots(3, 2, figsize=(10, 15))
ax1, ax2, ax3, ax4, ax5, ax6 = axs.flatten()

ax1.scatter(pos_Wthermo[0, :, 0], pos_Wthermo[0, :, 1], label='Initial Positions', color='blue')
ax1.scatter(pos_Wthermo[-1, :, 0], pos_Wthermo[-1, :, 1], label='Final Positions', marker='x', color='red')
ax1.set_title('Initial and final positions', fontsize = 15)
ax1.set_ylabel('Without Thermostat', fontsize = 15)
ax1.legend()

for i in range(pos_Wthermo.shape[1]):
    ax2.plot(pos_Wthermo[:, i, 0], pos_Wthermo[:, i, 1], lw=.5)
ax2.set_title('particles path(VV)', fontsize = 15)

ax3.scatter(pos_VSthermo[0, :, 0], pos_VSthermo[0, :, 1], label='Initial Positions', color='blue')
ax3.scatter(pos_VSthermo[-1, :, 0], pos_VSthermo[-1, :, 1], label='Final Positions', marker='x', color='red')
ax3.set_ylabel('Velocity Scaling Thermostat(VV)', fontsize = 15)
ax3.legend()

for i in range(pos_VSthermo.shape[1]):
    ax4.plot(pos_VSthermo[:, i, 0], pos_VSthermo[:, i, 1], lw=.5)

ax5.scatter(pos_Bthermo[0, :, 0], pos_Bthermo[0, :, 1], label='Initial Positions', color='blue')
ax5.scatter(pos_Bthermo[-1, :, 0], pos_Bthermo[-1, :, 1], label='Final Positions', marker='x', color='red')
ax5.set_ylabel('Berendenson Thermostat(VV)', fontsize = 15)
ax5.legend()

for i in range(pos_Bthermo.shape[1]):
    ax6.plot(pos_Bthermo[:, i, 0], pos_Bthermo[:, i, 1], lw=.5)    
```

```python
# energy of system for thermostats

import numpy as np
import matplotlib.pyplot as plt

kinetic_none = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/N_particle_KineticEnergyMD.dat'), axis = 1)
potential_none = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/N_particle_PotentialEnergyMD.dat'), axis = 1)
total_none = kinetic_none + potential_none

kinetic_vs = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/velocityScaling/N_particle_KineticEnergyMD.dat'), axis = 1)
potential_vs = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/velocityScaling/N_particle_PotentialEnergyMD.dat'), axis = 1)
total_vs = kinetic_vs + potential_vs

kinetic_br = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/berendensen/N_particle_KineticEnergyMD.dat'), axis = 1)
potential_br = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/thermostat/berendensen/N_particle_PotentialEnergyMD.dat'), axis = 1)
total_br = kinetic_br + potential_br


fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 18))

ax1.plot(kinetic_none, label=f'no thermostat')
ax1.plot(kinetic_vs, label=f'velocity scaling thermostat')
ax1.plot(kinetic_br, label=f'berendenson thermostat')
ax1.set_xlabel('Time')
ax1.set_ylabel('Energy')
ax1.set_title('Kinetic Energy of system', fontsize = 15)
ax1.legend()
ax1.grid()

ax2.plot(potential_none, label=f'no thermostat')
ax2.plot(potential_vs, label=f'velocity scaling thermostat')
ax2.plot(potential_br, label=f'berendenson thermostat')
ax2.set_xlabel('Time')
ax2.set_ylabel('Energy')
ax2.set_title('Potential Energy of system', fontsize = 15)
ax2.legend()
ax2.grid()

ax3.plot(total_none, label=f'no thermostat')
ax3.plot(total_vs, label=f'velocity scaling thermostat')
ax3.plot(total_br, label=f'berendenson thermostat')
ax3.set_xlabel('Time')
ax3.set_ylabel('Energy')
ax3.set_title('Total Energy of system', fontsize = 15)
ax3.legend()
ax3.grid()
```
