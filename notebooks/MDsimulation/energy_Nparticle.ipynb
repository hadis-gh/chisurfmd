```python
import numpy as np
import matplotlib.pyplot as plt

def readPositions(filename):
    data = np.loadtxt(filename)
    numSteps = data.shape[0]
    numParticles = data.shape[1] // 2
    positions = data.reshape(numSteps, numParticles, 2)
    return positions

path_euler = "/home/hadis/custom_vector/build/test/euler/N_particle_PosMD.dat"
pos_euler = readPositions(path_euler)

path_vv = "/home/hadis/custom_vector/build/test/velocityVerlet/N_particle_PosMD.dat"
pos_vv = readPositions(path_vv)

fig, axs = plt.subplots(2, 2, figsize=(10, 10))
ax1, ax2, ax3, ax4 = axs.flatten()

ax1.scatter(pos_euler[0, :, 0], pos_euler[0, :, 1], label='Initial Positions', color='blue')
ax1.scatter(pos_euler[-1, :, 0], pos_euler[-1, :, 1], label='Final Positions', marker='x', color='red')
ax1.set_title('Initial and final positions', fontsize = 15)
ax1.set_ylabel('Euler Integration', fontsize = 15)
ax1.legend()

for i in range(pos_euler.shape[1]):
    ax2.plot(pos_euler[:, i, 0], pos_euler[:, i, 1], lw=.5)
ax2.set_title('particles path', fontsize = 15)

ax3.scatter(pos_vv[0, :, 0], pos_vv[0, :, 1], label='Initial Positions', color='blue')
ax3.scatter(pos_vv[-1, :, 0], pos_vv[-1, :, 1], label='Final Positions', marker='x', color='red')
ax3.set_ylabel('Velocity Verlet Integration', fontsize = 15)
ax3.legend()

for i in range(pos_vv.shape[1]):
    ax4.plot(pos_vv[:, i, 0], pos_vv[:, i, 1], lw=.5)
```

```python
import numpy as np
import matplotlib.pyplot as plt

kinetic_euler = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/euler/N_particle_KineticEnergyMD.dat'), axis = 1)
potential_euler = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/euler/N_particle_PotentialEnergyMD.dat'), axis = 1)
total_euler = kinetic_euler + potential_euler

kinetic_vv = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/N_particle_KineticEnergyMD.dat'), axis = 1)
potential_vv = np.sum(np.loadtxt('/home/hadis/custom_vector/build/test/velocityVerlet/N_particle_PotentialEnergyMD.dat'), axis = 1)
total_vv = kinetic_vv + potential_vv

fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 18))

ax1.plot(kinetic_euler, label=f'Euler')
ax1.plot(kinetic_vv, label=f'Velocity Verlet')
# ax1.plot(total_vv, label=f'Total vv')
ax1.set_xlabel('Time')
ax1.set_ylabel('Energy')
ax1.set_title('Kinetic Energy of system', fontsize = 15)
ax1.legend()

ax2.plot(potential_euler, label=f'Euler')
ax2.plot(potential_vv, label=f'Velocity Verlet')
# ax2.plot(total_vv, label=f'total vv')

ax2.set_xlabel('Time')
ax2.set_ylabel('Energy')
ax2.set_title('Potential Energy of system', fontsize = 15)
ax2.legend()

ax3.plot(total_euler, label=f'Euler')
ax3.plot(total_vv, label=f'Velocity Verlet')
ax3.set_xlabel('Time')
ax3.set_ylabel('Energy')
ax3.set_title('Total Energy of system', fontsize = 15)
ax3.legend()
```
