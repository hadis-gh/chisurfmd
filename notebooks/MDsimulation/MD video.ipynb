```python
import os
import numpy as np
import matplotlib.pyplot as plt
import cv2
from glob import glob

cooling_dir = "/home/hadis/custom_vector/buildVS/Oct_Dot_100_50/cooling_configs/"
heating_dir = "/home/hadis/custom_vector/buildVS/Oct_Dot_100_50/heating_configs/"
output_dir = "./frames/"
os.makedirs(output_dir, exist_ok=True)

def save_positions_as_image(data_file, frame_num):
    data = np.loadtxt(data_file)
    x = data[:, 0]
    y = data[:, 1]
    
    plt.figure(figsize=(6, 6))
    plt.scatter(x, y, c='blue', marker='o', s=20)
    plt.xlim(-1, 21)
    plt.ylim(-1, 21)
    plt.title(f"Frame {frame_num}")
    plt.xlabel("X Position")
    plt.ylabel("Y Position")
    plt.grid(True)
    plt.savefig(f"{output_dir}/frame_{frame_num:04d}.png")
    plt.close()

cooling_files = sorted(glob(f"{cooling_dir}/N_Config_*.dat"), reverse=True)
for i, file in enumerate(cooling_files):
    save_positions_as_image(file, i)

heating_files = sorted(glob(f"{heating_dir}/N_Config_*.dat"))
for i, file in enumerate(heating_files):
    save_positions_as_image(file, i + len(cooling_files))

def create_video_from_images(output_dir, output_video):
    images = sorted(glob(f"{output_dir}/*.png"))
    frame = cv2.imread(images[0])
    height, width, _ = frame.shape
    video = cv2.VideoWriter(output_video, cv2.VideoWriter_fourcc(*'mp4v'), 30, (width, height))
    
    for image in images:
        video.write(cv2.imread(image))
    
    video.release()

output_video = "./particle_motion.mp4"
create_video_from_images(output_dir, output_video)

print(f"Video saved as {output_video}")
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import math

file_path = "/home/hadis/custom_vector/buildParticleOriented/buildVS/11Nov_velver_100_100/N_particle_PosMD.dat"

skip_rows = 50
frame_size = 800
particle_radius = 9
output_file = 'particle_movement2_phi.mp4'

positions = []
with open(file_path, 'r') as file:
    for i, line in enumerate(file):
        if i % skip_rows == 0:
            data = line.strip().split()
            data = list(map(float, data[1:]))
            positions.append(data)
positions = np.array(positions)

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_file, fourcc, 30.0, (frame_size + 100, frame_size))  # Increase frame width to accommodate thermometer

# Update scale and offset to match the real box dimensions
scale = frame_size / 20  # Scale factor to fit positions on frame
offset = 0  # No need to add offset since the range is already from 0 to 20

def map_to_frame(x, y):
    return int(x * scale + offset), int(y * scale + offset)

# Function to draw thermometer
def draw_thermometer(frame, current_temp, max_temp=1.0):
    # Thermometer dimensions
    thermo_x = frame_size + 20
    thermo_y_start = 50
    thermo_y_end = frame_size - 50
    thermo_width = 30

    # Draw the gradient
    for y in range(thermo_y_start, thermo_y_end):
        ratio = (y - thermo_y_start) / (thermo_y_end - thermo_y_start)
        color = (int(255 * ratio), 0, int(255 * (1 - ratio)))  # Blue to red gradient
        cv2.line(frame, (thermo_x, y), (thermo_x + thermo_width, y), color, 1)

    # Draw the current temperature arrow
    arrow_y = int(thermo_y_end - (current_temp / max_temp) * (thermo_y_end - thermo_y_start))
    cv2.arrowedLine(frame, (thermo_x + thermo_width + 10, arrow_y), (thermo_x + thermo_width + 30, arrow_y), (0, 0, 0), 2)

# Function to draw a box around particles
def draw_particle_box(frame):
    box_top_left = (0, 0)
    box_bottom_right = (frame_size, frame_size)
    cv2.rectangle(frame, box_top_left, box_bottom_right, (0, 0, 0), 2)  # Black box with thickness 2

# Function to draw particles with orientation
def draw_particles_with_orientation(frame, x, y, phi, radius, color):
    # Draw particle as a circle
    cx, cy = map_to_frame(x, y)
    cv2.circle(frame, (cx, cy), radius, color, -1)
    
    # Draw orientation line
    line_length = radius * 2
    end_x = int(cx + line_length * math.cos(math.radians(phi)))
    end_y = int(cy - line_length * math.sin(math.radians(phi)))
    cv2.line(frame, (cx, cy), (end_x, end_y), (0, 0, 0), 2)  # Black line to indicate orientation

for i, frame_data in enumerate(positions):
    frame = np.ones((frame_size, frame_size + 100, 3), dtype=np.uint8) * 255  # Extend frame width to fit thermometer

    # Draw particles with orientation
    for j in range(0, len(frame_data), 3):
        x, y, phi = frame_data[j], frame_data[j+1], frame_data[j+2]
        color = (100, 100, 100)
        draw_particles_with_orientation(frame, x, y, phi, particle_radius, color)

    # Draw box around particles
    draw_particle_box(frame)

    # Calculate temperature (linearly varying from 1 to 0 and back to 1)
    total_frames = len(positions)
    half_frames = total_frames // 2
    if i < half_frames:
        current_temp = 1 - (i / half_frames)
    else:
        current_temp = (i - half_frames) / half_frames

    # Draw thermometer
    draw_thermometer(frame, current_temp)

    out.write(frame)
    
    if i % 50 == 0:
        print(f"Processing frame {i}/{len(positions)}")

out.release()
print("Video saved as", output_file) 

plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
plt.title("Particle Movement at Last Frame")
plt.show()
```
