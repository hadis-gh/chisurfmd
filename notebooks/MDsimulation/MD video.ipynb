```python
import os
import numpy as np
import matplotlib.pyplot as plt
import cv2
from glob import glob

cooling_dir = "/home/hadis/custom_vector/buildVS/Oct_Dot_100_50/cooling_configs/"
heating_dir = "/home/hadis/custom_vector/buildVS/Oct_Dot_100_50/heating_configs/"
output_dir = "./frames/"
os.makedirs(output_dir, exist_ok=True)

def save_positions_as_image(data_file, frame_num):
    data = np.loadtxt(data_file)
    x = data[:, 0]
    y = data[:, 1]
    
    plt.figure(figsize=(6, 6))
    plt.scatter(x, y, c='blue', marker='o', s=20)
    plt.xlim(-1, 21)
    plt.ylim(-1, 21)
    plt.title(f"Frame {frame_num}")
    plt.xlabel("X Position")
    plt.ylabel("Y Position")
    plt.grid(True)
    plt.savefig(f"{output_dir}/frame_{frame_num:04d}.png")
    plt.close()

cooling_files = sorted(glob(f"{cooling_dir}/N_Config_*.dat"), reverse=True)
for i, file in enumerate(cooling_files):
    save_positions_as_image(file, i)

heating_files = sorted(glob(f"{heating_dir}/N_Config_*.dat"))
for i, file in enumerate(heating_files):
    save_positions_as_image(file, i + len(cooling_files))

def create_video_from_images(output_dir, output_video):
    images = sorted(glob(f"{output_dir}/*.png"))
    frame = cv2.imread(images[0])
    height, width, _ = frame.shape
    video = cv2.VideoWriter(output_video, cv2.VideoWriter_fourcc(*'mp4v'), 30, (width, height))
    
    for image in images:
        video.write(cv2.imread(image))
    
    video.release()

output_video = "./particle_motion.mp4"
create_video_from_images(output_dir, output_video)

print(f"Video saved as {output_video}")
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import cv2

data_file = "/home/hadis/custom_vector/buildVS/nv_100_20/N_particle_PosMD.dat"
output_dir = "./frames_md/"
os.makedirs(output_dir, exist_ok=True)

def save_positions_as_image(positions, frame_num):
    plt.figure(figsize=(6, 6))
    x = positions[0::2]
    y = positions[1::2]
    plt.scatter(x, y, c='blue', marker='o', s=20)
    plt.xlim(-1, 21)
    plt.ylim(-1, 21)
    plt.title(f"Frame {frame_num}")
    plt.xlabel("X Position")
    plt.ylabel("Y Position")
    plt.grid(True)
    plt.savefig(f"{output_dir}/frame_{frame_num:04d}.png")
    plt.close()

data = np.loadtxt(data_file)
num_particles = (data.shape[1] - 1) // 2
for i, row in enumerate(data):
    positions = row[1:]
    save_positions_as_image(positions, i)

def create_video_from_images(output_dir, output_video):
    images = sorted([img for img in os.listdir(output_dir) if img.endswith(".png")])
    frame = cv2.imread(os.path.join(output_dir, images[0]))
    height, width, _ = frame.shape
    video = cv2.VideoWriter(output_video, cv2.VideoWriter_fourcc(*'mp4v'), 10, (width, height))
    
    for image in images:
        video.write(cv2.imread(os.path.join(output_dir, image)))
    
    video.release()

output_video = "./particle_motion_md.mp4"
create_video_from_images(output_dir, output_video)

print(f"Video saved as {output_video}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

data = np.loadtxt("/home/hadis/custom_vector/buildParticleDot/buildVS/Oct30_euler_100_100/N_particle_PosMD.dat")[:, 1:]
data = data[::100]

degrees_freedom = 3
n_steps = data.shape[0]
n_particles = int(data.shape[1] / degrees_freedom)

positions = data.reshape(n_steps, n_particles, degrees_freedom)

fig, ax = plt.subplots()
scat = ax.scatter(positions[0, :, 0], positions[0, :, 1], s=10, c='blue', alpha=0.6)

ax.set_xlim(positions[:, :, 0].min(), positions[:, :, 0].max())
ax.set_ylim(positions[:, :, 1].min(), positions[:, :, 1].max())

plt.gca().set_aspect('equal', adjustable='box')
plt.xlim(0, 20)
plt.ylim(0, 20)
ax.set_xlabel('X position')
ax.set_ylabel('Y position')
ax.set_title('Particle Movements')

def update(frame):
    scat.set_offsets(positions[frame])
    return scat,

ani = FuncAnimation(fig, update, frames=n_steps, interval=50, blit=True)
# ani.save('particle_movement.mp4', writer='ffmpeg', fps=10)
ani.save('31oct_dot_euler.gif', writer='pillow', fps=10)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

data = np.loadtxt("/home/hadis/custom_vector/buildParticleOriented/buildVS/28oct_100_100/N_particle_PosMD.dat")[:, 1:]
data = data[::100]

degrees_freedom = 3
n_steps = data.shape[0]
n_particles = int(data.shape[1] / degrees_freedom)

positions = data.reshape(n_steps, n_particles, degrees_freedom)

fig, ax = plt.subplots()
scat = ax.scatter(positions[0, :, 0], positions[0, :, 1], s=10, c='blue', alpha=0.6)

ax.set_xlim(positions[:, :, 0].min(), positions[:, :, 0].max())
ax.set_ylim(positions[:, :, 1].min(), positions[:, :, 1].max())

plt.gca().set_aspect('equal', adjustable='box')
plt.xlim(0, 20)
plt.ylim(0, 20)
ax.set_xlabel('X position')
ax.set_ylabel('Y position')
ax.set_title('Particle Movements')

def update(frame):
    scat.set_offsets(positions[frame, :, :2])
    scat.set_array(positions[frame, :, 2])  # Update colors based on orientation
    return

ani = FuncAnimation(fig, update, frames=n_steps, interval=50, blit=True)
# ani.save('particle_movement.mp4', writer='ffmpeg', fps=10)
ani.save('particle_movement.gif', writer='pillow', fps=10)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib import cm

data = np.loadtxt("/home/hadis/custom_vector/buildParticleOriented/buildVS/11Nov_100_100/N_particle_PosMD.dat")[:, 1:]
data = data[::30]

degrees_freedom = 3
n_steps = data.shape[0]
n_particles = int(data.shape[1] / degrees_freedom)
positions = data.reshape(n_steps, n_particles, degrees_freedom)

orientations = positions[:, :, 2]
colors = cm.viridis((orientations - orientations.min()) / (orientations.max() - orientations.min()))

fig, ax = plt.subplots()
scat = ax.scatter(positions[0, :, 0], positions[0, :, 1], s=10, c=colors[0], alpha=0.6)

ax.set_xlim(positions[:, :, 0].min(), positions[:, :, 0].max())
ax.set_ylim(positions[:, :, 1].min(), positions[:, :, 1].max())
plt.gca().set_aspect('equal', adjustable='box')
ax.set_xlabel('X position')
ax.set_ylabel('Y position')
ax.set_title('Particle Movements with Orientation Coloring')

def update(frame):
    scat.set_offsets(positions[frame, :, :2])
    scat.set_color(colors[frame])
    return scat,

ani = FuncAnimation(fig, update, frames=n_steps, interval=50, blit=True)

ani.save('particle_movement_orientation.gif', writer='pillow', fps=10)
plt.show()
```
