```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize

def draw_circles(radius, length, centers, colors):
    fig, ax = plt.subplots()

    ax.set_xlim(0, length)
    ax.set_ylim(0, length)

    for i, center in enumerate(centers):
        circle = plt.Circle(center, radius, color=colors[i], fill=True)
        ax.add_artist(circle)

    ax.set_aspect('equal', adjustable='box')
    plt.show()

def converter_func(s):
    return float(s.decode('utf-8').strip('()'))

def read_centers_numpy(file_name):
    return np.loadtxt(file_name, delimiter=',', converters={0: converter_func, 1: converter_func})

circle_radius = 1
length = 50

file_name = '/home/hadis/selfassembly/build/test/circlesDLAtest.txt'
circle_centers = read_centers_numpy(file_name)

cmap = plt.cm.get_cmap('magma', len(circle_centers))
colors = [cmap(i) for i in range(len(circle_centers))]

draw_circles(circle_radius, length, circle_centers, colors)
```
