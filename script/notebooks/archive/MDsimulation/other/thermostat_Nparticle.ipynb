```python
import numpy as np
import matplotlib.pyplot as plt

def readPositions(filename):
    data = np.loadtxt(filename)[:, 1:]
    numSteps = data.shape[0]
    numParticles = data.shape[1] // 2
    positions = data.reshape(numSteps, numParticles, 2)
    return positions

def plotTraj(inputs, labels):
    num_plots = len(inputs)
    
    fig, axs = plt.subplots(num_plots, 2, figsize=(10, 5 * num_plots))
    
    if num_plots == 1:
        axs = np.array([axs])

    for j, fname in enumerate(inputs):
        label = labels[j]
        pos = readPositions(fname)

        axs[j][0].scatter(pos[0, :, 0], pos[0, :, 1], label='Initial Positions', color='blue')
        axs[j][0].scatter(pos[-1, :, 0], pos[-1, :, 1], label='Final Positions', marker='x', color='red')
        axs[j][0].set_title('Initial and final positions', fontsize = 15)
        axs[j][0].set_ylabel(label, fontsize = 15)
        axs[j][0].legend()

        for i in range(pos.shape[1]):
            axs[j][1].plot(pos[:, i, 0], pos[:, i, 1], lw=.5)
        axs[j][1].set_title(label, fontsize = 15)
    
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def load_and_sum(file_path):
    data = np.loadtxt(file_path)
    return np.sum(data, axis=1)

def calculate_energies(base_path, subpath=""):
    kinetic = load_and_sum(f"{base_path}{subpath}N_particle_KineticEnergyMD.dat")
    potential = load_and_sum(f"{base_path}{subpath}N_particle_PotentialEnergyMD.dat")
    total = kinetic + potential
    return kinetic, potential, total

def plot_energy(ax, energies, labels, title):
    for j, energy in enumerate(energies):
        ax.plot(energy, label=labels[j])
    ax.set_xlabel("Time")
    ax.set_ylabel("Energy")
    ax.set_title(title, fontsize=15)
    ax.legend()
    ax.grid()

base_path = '/home/hadis/selfassembly/build/test/velocityVerlet/'

thermotypes_labels = ['no thermostat', 'velocity scaling thermostat', 'berendsen thermostat', 'andersen thermostat']
energies_labels = ['Kinetic Energy', 'Potential Energy', 'Total Energy']

energies = [
    calculate_energies(base_path),
    calculate_energies(base_path, 'thermostat/velocityScaling/'),
    calculate_energies(base_path, 'thermostat/berendsen/'),
    calculate_energies(base_path, 'thermostat/andersen/')
]

fig, ax = plt.subplots(3, 1, figsize=(10, 18))

for j, energy_label in enumerate(energies_labels):
    energy_data = [energy[j] for energy in energies]
    plot_energy(ax[j], energy_data, thermotypes_labels, energy_label)

plt.tight_layout()
plt.show()
```
