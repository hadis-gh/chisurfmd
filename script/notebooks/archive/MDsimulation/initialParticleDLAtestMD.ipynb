```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize

def draw_circles(radius, areaL, centers, colors):
    fig, ax = plt.subplots()


    for i, center in enumerate(centers):
        circle = plt.Circle(center, radius, color=colors[i], fill=True)
        ax.add_artist(circle)

    ax.set_aspect('equal', adjustable='box')
    ax.set_xlim(np.min(centers[:, 0]) - 2 * radius, np.max(centers[:, 0]) + 2 * radius)
    ax.set_ylim(np.min(centers[:, 1]) - 2 * radius, np.max(centers[:, 1]) + 2 * radius)
    plt.show()

areaL = 10

particle = np.loadtxt("/home/hadis/selfassembly/build/test/checkInit.dat")
particle_radius = particle[0, 2]
particle_centers = particle[:, :2]

cmap = plt.cm.get_cmap('magma', len(particle_centers))
colors = [cmap(i) for i in range(len(particle_centers))]

draw_circles(particle_radius, areaL, particle_centers, colors)
particle.shape
```

```python
import numpy as np
import matplotlib.pyplot as plt

kinetic = np.sum(np.loadtxt('/home/hadis/selfassembly/build/test/velocityVerlet/N_particle_KineticEnergyMD.dat'), axis = 1)
potential = np.sum(np.loadtxt('/home/hadis/selfassembly/build/test/velocityVerlet/N_particle_PotentialEnergyMD.dat'), axis = 1)
total = kinetic + potential

fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 18))

ax1.plot(kinetic)
ax1.set_xlabel('Time')
ax1.set_ylabel('Energy')
ax1.set_title('Kinetic Energy of system', fontsize = 15)
ax1.grid()

ax2.plot(potential)
ax2.set_xlabel('Time')
ax2.set_ylabel('Energy')
ax2.set_title('Potential Energy of system', fontsize = 15)
ax2.grid()

ax3.plot(total)
ax3.set_xlabel('Time')
ax3.set_ylabel('Energy')
ax3.set_title('Total Energy of system', fontsize = 15)
ax3.grid()
```
